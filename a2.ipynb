```python
import pandas as pd 
df = pd.read_csv('digikala_text.csv')
df.head()

x = df['Text']
y = df['Suggestion']

from sklearn.feature_extraction.text import TfidfVectorizer
tarjome = TfidfVectorizer()
x_number = tarjome.fit_transform(x)

from sklearn.model_selection import train_test_split
x_train , x_test , y_train , y_test = train_test_split(x_number , y , test_size=0.1)

from sklearn.svm import SVC 
model = SVC(kernel='linear')
model.fit(x_train , y_train)

from sklearn.metrics import accuracy_score
y_exam = model.predict(x_test)
accuracy_score(y_test , y_exam)

test = ["خیلی راضی بودم"]

test_number = tarjome.transform(test)
javab = model.predict(test_number)

if javab == 1 :
    print('مثبت')
else :
    print("منفی")
```
